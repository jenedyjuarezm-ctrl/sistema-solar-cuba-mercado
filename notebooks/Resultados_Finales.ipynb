## Gráfica 1: Composición Metodológica del Dataset

## Gráfica 2: La Crisis Energética y la Paradoja FV (ONEI)
*(Muestra cómo cayó la generación total, subió la potencia FV pero cayó su generación)*

```python
data_onei = {
    'Año': [2020, 2021, 2022],
    'Generación Total (GW·h)': [19070.9, 17965.5, 15732.1],
    'Generación FV (GW·h)': [260.8, 265.0, 241.5],
    'Potencia FV (MW)': [221.5, 241.8, 263.0]
}
df_onei = pd.DataFrame(data_onei)

fig, ax1 = plt.subplots(figsize=(12, 6))

# Eje 1: Generación Total
color = '#1f77b4'
ax1.set_xlabel('Año', fontweight='bold')
ax1.set_ylabel('Generación Total (GW·h)', color=color, fontweight='bold')
ax1.plot(df_onei['Año'], df_onei['Generación Total (GW·h)'], marker='o', color=color, linewidth=3, markersize=10, label='Generación Total')
ax1.fill_between(df_onei['Año'], df_onei['Generación Total (GW·h)'], color=color, alpha=0.1)
ax1.tick_params(axis='y', labelcolor=color)
ax1.set_ylim(13000, 21000)

# Eje 2: Generación FV y Potencia FV
ax2 = ax1.twinx()  
color_fic = '#2ca02c'
color_pot = '#d62728'

ax2.plot(df_onei['Año'], df_onei['Generación FV (GW·h)'], marker='s', color=color_fic, linewidth=3, markersize=10, label='Generación FV')
ax2.plot(df_onei['Año'], df_onei['Potencia FV (MW)'], marker='^', color=color_pot, linewidth=3, markersize=10, linestyle='--', label='Potencia Instalada FV')
ax2.set_ylabel('Generación FV (GW·h) / Potencia (MW)', fontweight='bold')
ax2.tick_params(axis='y')
ax2.set_ylim(180, 300)

# Leyenda combinada
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3, frameon=False)

plt.title('Cuba (2020-2022): Caída del Sistema Eléctrico Total\nvs. La Paradoja de la Eficiencia Fotovoltaica')
plt.xticks(df_onei['Año'])
fig.tight_layout()
plt.show()
```

## Gráfica 3: Inversión por Sector
*(El impacto del esfuerzo ciudadano vs la infraestructura)*

```python
sectores = ['Residencias\n(Primarias)', 'Centros / Locales\n(Web)', 'Solineras\n(Web)']
potencia_mediana = [2.0, 4.5, 40.0]
colores = ['#2ca02c', '#d62728', '#9467bd']

plt.figure(figsize=(10, 6))
bars = plt.bar(sectores, potencia_mediana, color=colores, width=0.5, edgecolor='black')

plt.title('Potencia Instalada Mediana por Tipo de Entidad')
plt.ylabel('Potencia Mediana (kWp)', fontweight='bold')
plt.ylim(0, 48)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1.5, f'{yval} kWp', ha='center', va='bottom', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```

## Gráfica 4: El Impacto Climático (NASA POWER)
*(La paradoja del verano: mucho sol, pero mucho calor)*

```python
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
irradiancia = [3.8, 4.8, 5.9, 6.6, 6.7, 6.1, 6.6, 6.2, 5.7, 4.7, 3.8, 3.4]
temperatura = [23.2, 24.0, 25.6, 27.1, 28.1, 28.2, 28.6, 28.5, 28.1, 26.8, 24.8, 23.3]

fig, ax1 = plt.subplots(figsize=(12, 6))
color = '#bcbd22'
ax1.set_xlabel('Mes', fontweight='bold')
ax1.set_ylabel('Irradiancia (kWh/m²/día)', color=color, fontweight='bold')
ax1.bar(meses, irradiancia, color=color, alpha=0.7, label='Irradiancia')
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = '#e377c2'
ax2.set_ylabel('Temperatura Promedio (°C)', color=color, fontweight='bold')
ax2.plot(meses, temperatura, color=color, marker='o', linewidth=3, markersize=8, label='Temperatura')
ax2.tick_params(axis='y', labelcolor=color)

plt.annotate('Alta Irradiancia, pero Calor Extremo\n(Pérdida de eficiencia en verano)', 
             xy=('Ago', 28.5), xytext=('Sep', 31),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
             fontsize=12, ha='center', fontweight='bold', bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="black", lw=1))

plt.title('La Paradoja del Verano en Cuba: Máxima Irradiancia vs. Temperatura')
fig.tight_layout()
plt.show()
```

## Gráfica 5: Análisis Cualitativo (Motivaciones)
*(¿Por qué los cubanos compran paneles solares?)*

```python
motivos = ['Cortes de Energía\n(Supervivencia)', 'Ahorro Económico', 'Conciencia Ambiental']
menciones = [32, 5, 2]

plt.figure(figsize=(10, 5))
bars = plt.barh(motivos, menciones, color=['#d62728', '#1f77b4', '#2ca02c'], height=0.5, edgecolor='black')
plt.title('¿Por qué los cubanos invierten en energía solar?')
plt.xlabel('Número de Menciones (Muestra total: 32 residencias)', fontweight='bold')
plt.gca().invert_yaxis()

for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, f'{width} menciones', ha='left', va='center', fontweight='bold', fontsize=12)

plt.xlim(0, 38)
plt.tight_layout()
plt.show()
```
