# 06 — Visualizaciones para el Storytelling

Gráficos clave para la presentación final y el reporte.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA = Path.cwd().parent / 'data'
df = pd.read_parquet(DATA / 'clean' / 'encuestas_consolidadas.parquet')
sns.set_theme(style='whitegrid')
```

## 1. Distribución de la muestra (El impacto de la investigación)

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y='tipo_entidad', palette='viridis', order=df['tipo_entidad'].value_counts().index)
plt.title('Total de Entidades Documentadas (Primarias + Web)', fontsize=14)
plt.xlabel('Cantidad de registros')
plt.ylabel('Tipo de Entidad')
plt.show()
```

## 2. Potencia mediana por sector (Los locales invierten más)

```python
# Filtrar solo residentes y centros para ver la diferencia
df_comp = df[df['tipo_entidad'].isin(['residente', 'centro', 'centro_web'])]
plt.figure(figsize=(8, 5))
sns.barplot(data=df_comp, x='tipo_entidad', y='potencia_pico_kwp', estimator='median', palette='Set2', errorbar=None)
plt.title('Mediana de Potencia Instalada (kWp) por Sector')
plt.ylabel('Potencia mediana (kWp)')
plt.show()
```
