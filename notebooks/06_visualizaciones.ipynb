# 06 — Visualizaciones

**Librerias usadas:** pandas, matplotlib, seaborn

```python
import sys, os
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'solineras'))
os.chdir(ROOT)
DATA = ROOT / 'data'

p = DATA / 'raw' / 'externas' / 'onei_anuario.csv'
if not p.exists():
    from extraer_onei import extraer
    extraer()
onei = pd.read_csv(p)
sns.set_theme(style='whitegrid')
```

## 1. Area

```python
s = onei[onei['indicador'] == 'eolica_fotovoltaica_gwh'].sort_values('anio')
plt.figure(figsize=(10, 5))
plt.fill_between(s['anio'], s['valor'], alpha=0.4, color='orange')
plt.plot(s['anio'], s['valor'], marker='o', color='darkorange', linewidth=2)
plt.title('Generacion FV + eolica 2020-2024')
plt.ylabel('GWh')
plt.show()
```

## 2. Torta consumo 2024

```python
comp = onei[(onei['anio'] == 2024) & (onei['cuadro'] == '10.19')]
comp = comp[comp['indicador'].isin(['estatal_gwh', 'poblacion_gwh', 'perdidas_gwh'])]
if not comp.empty:
    plt.figure(figsize=(7, 7))
    plt.pie(comp['valor'], labels=comp['indicador'], autopct='%1.1f%%', startangle=90)
    plt.title('Composicion del consumo electrico 2024')
    plt.show()
```

## 3. Barras

```python
piv = onei[onei['indicador'].isin(['hidroelectricas_gwh', 'eolica_fotovoltaica_gwh'])]
if not piv.empty:
    piv = piv.pivot_table(index='anio', columns='indicador', values='valor')
    piv.plot(kind='bar', figsize=(10, 5))
    plt.title('Generacion renovable por fuente')
    plt.ylabel('GWh')
    plt.xticks(rotation=0)
    plt.show()
```
