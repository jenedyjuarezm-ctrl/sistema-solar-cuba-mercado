# 10 — Storytelling: Las Tablas del Reporte

Generación de tablas de contexto para el informe final.

```python
import pandas as pd
from pathlib import Path

DATA = Path.cwd().parent / 'data'
df = pd.read_parquet(DATA / 'clean' / 'encuestas_consolidadas.parquet')

# Tabla 1: Resumen por tipo de entidad
resumen = df.groupby('tipo_entidad').agg(
    registros=('id', 'count'),
    potencia_mediana_kwp=('potencia_pico_kwp', 'median'),
    bateria_mediana_kwh=('capacidad_baterias_kwh', 'median'),
    satisfaccion_media=('satisfaccion', 'mean')
).round(2)

print('TABLA 1: Resumen de la Muestra')
display(resumen)
```

## Tabla 2: Top 5 Inversiones (El impacto económico)

```python
top_inversiones = df[df['costo_instalacion'].notna()].sort_values('costo_instalacion', ascending=False)
cols = ['id', 'tipo_entidad', 'potencia_pico_kwp', 'costo_instalacion', 'moneda_costo']
display(top_inversiones[cols].head(5))
```
