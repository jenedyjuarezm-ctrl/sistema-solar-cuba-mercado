Verificamos la existencia de todas las fuentes (Primarias y Secundarias).

```python
import sys
import os
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'solineras'))
os.chdir(ROOT)

from auditoria_fuentes import auditar

# 1. Ejecutar auditoría de fuentes
cobertura = auditar()

# 2. Mostrar tabla resumen
resumen = pd.DataFrame([
    {'fuente': k, 'n': v['n'], 'cumple': v['cumple'], 'faltan': v['faltan']}
    for k, v in cobertura['fuentes'].items()
])
display(resumen)
```
